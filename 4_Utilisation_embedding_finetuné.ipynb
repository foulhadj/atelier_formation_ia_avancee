# Comment utiliser le modèle finetuné ?


### importation des librairies

```python
from sentence_transformers import SentenceTransformer
from datasets import load_dataset
import numpy as np
```

### Charger le modèle finetuné

```python
model = SentenceTransformer("Fe2x/distilroberta-ai-job-embeddings")
```

### Charger les données de HF

```python
dataset = load_dataset("Fe2x/ai-job-embedding-finetuning")
```

### Utilisez le!

```python
# Nouvelle demande utilisateur
query = "data scienist 6 year experience, LLMs, credit risk, content marketing"
query_embedding = model.encode(query)

# Encodage des descriptions de poste
jd_embeddings = model.encode(dataset["test"]["job_description_pos"])

# Calculer les similarités sémantiques
similarities = model.similarity(query_embedding, jd_embeddings)
print(similarities.shape)
```

```python
# Trier les descriptions de postes en fonction de leur similitude
simiarities_argsorted = np.argsort(similarities.numpy(), axis=1)

# Retourner la description de poste la plus similaire
dataset["test"]["job_description_pos"][int(simiarities_argsorted[0][-1])]
```
